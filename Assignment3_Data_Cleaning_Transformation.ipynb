# Phase 1 – Data Acquisition & Consolidation

The Online Retail II dataset is distributed across multiple worksheets. To perform consistent data cleaning and analysis, all worksheets were loaded and merged into a single DataFrame. The index was reset during concatenation to avoid duplicate row indices.

```python
import pandas as pd
import numpy as np

xls = pd.ExcelFile("online_retail_II.xlsx")

print(xls.sheet_names)

dfs = []

for sheet in xls.sheet_names:
    temp = pd.read_excel(xls, sheet_name=sheet)
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)

print(df.shape)

df.head()
```

# Phase 7

```python
df["LineTotal"] = df["Quantity"] * df["Price"]
```

# Phase 8

```python
reference_date = df["InvoiceDate"].max()

customer_summary = (
    df.groupby("Customer ID")
      .agg(
          Monetary=("LineTotal","sum"),
          Frequency=("Invoice","nunique"),
          LastPurchase=("InvoiceDate","max")
      )
)

customer_summary["Recency"] = (
    reference_date -
    customer_summary["LastPurchase"]
).dt.days

customer_summary.drop(columns="LastPurchase", inplace=True)

customer_summary.head()
```

```python
customer_summary.to_csv("customer_summary.csv")
```

**Question 1**
Why might a company store years of transactional data across multiple sheets instead of one?

**Answer**
Large transactional datasets are often split across multiple worksheets to improve organization and file management. Different sheets may represent different years or reporting periods, making the data easier to maintain and reducing the size of individual worksheets.

**Question 2**
What could go wrong if you combined the sheets incorrectly?

**Answer**
If the sheets were merged incorrectly, columns could become misaligned, duplicate rows could be introduced, or transactions could be lost. These errors would affect every later cleaning and analysis step.

**Question 3**
Why reset the index?

**Answer**
The index was reset after concatenation to create one continuous row numbering system. Keeping the original indices would result in duplicate index values from different worksheets, making later processing more difficult.

# Phase 5 Reflection

Transactions with missing Customer IDs were removed because the objective of this assignment is to create a customer-level summary. A missing Customer ID cannot be reliably inferred or imputed, since customer identifiers are unique labels rather than numerical values. Assigning a guessed ID would incorrectly attribute purchases to the wrong customer and produce misleading values for monetary, frequency, and recency measures. Therefore, these records were excluded from the analysis.

The data cleaning process focused on producing a reliable customer-level dataset suitable for behavioural analysis. Cancellation invoices, administrative stock codes, missing customer identifiers, and invalid prices were removed because they do not represent completed customer purchases. A LineTotal feature was engineered to calculate the monetary value of each transaction before aggregation. Customer-level metrics for Monetary, Frequency, and Recency were then calculated to summarize purchasing behaviour. Overall, each cleaning decision was guided by the specific analytical objective of understanding customer purchasing patterns rather than preserving every original transaction.
